### Go through charges and and make a new column with clean charges
While going through charges, also make a column for the charge statute
Then clean and and see all the unique charges
Note: Some people may have multiple charges, make a column for each charge


```python
import os 
import pandas as pd
```

```python
NOTEBOOK_DIR = os.getcwd()
DATA_DIR = os.path.join(NOTEBOOK_DIR, "..", "data")
CHECKPOINTS_DIR = os.path.join(DATA_DIR, "checkpoints")
DATA_PATH = os.path.join(CHECKPOINTS_DIR, "checkpoint12_age.csv")
OUT_DIR = os.path.join(DATA_DIR, "checkpoint13_cleaned_charges")
```

```python
cp12 = pd.read_csv(DATA_PATH)
mask = cp12['Charges'].notna() & (cp12['Charges'].astype(str).str.strip() != '')
cp12_charges = cp12[mask].copy().reset_index(drop=True)
```

```python
print(cp12_charges["Charges"].unique().tolist())
```

### Entries that are: "LAWRENCE, MA" or "LAWRENCE MA" or "LAWRENCE,MA" should be removed from the charges column

```python
import numpy as np

pattern = r"\bLAWRENCE,\s*MA\b"

affected_mask = cp12_charges["Charges"].astype(str).str.contains(pattern, case=False, na=False)
n_affected = int(affected_mask.sum())

# set those entries to missing, do not drop rows
cp12_charges.loc[affected_mask, "Charges"] = np.nan

print(f"Charges entries cleared: {n_affected}")
```

### Remove entries that are "LAWRENCE, MA" from Charges

```python
import re
import numpy as np

# Create a mask that finds "LAWRENCE , MA" (with irregular spacing, case-insensitive)
mask = cp12_charges['Charges'].astype(str).str.contains(
    r'LAWRENCE\s*,\s*MA',  # regex: "LAWRENCE" + optional spaces + comma + optional spaces + "MA"
    case=False, na=False
)

# Clear those entries without dropping the row
cp12_charges.loc[mask, 'Charges'] = np.nan

# Count how many were affected
print("Rows affected:", mask.sum())
```
