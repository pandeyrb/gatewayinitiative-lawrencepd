## Importing Packages

```python
import geopandas as gpd
from shapely import wkt
```

## Loading Data

```python
school_df = gpd.read_file('../../data/secondary_data/Lawrence_schools.csv')
```

```python
school_df.head()
```

```python
# Convert WKT text to shapely geometry objects
school_df["geometry"] = school_df["geometry"].apply(wkt.loads)

# Create GeoDataFrame with correct CRS
school_gdf = gpd.GeoDataFrame(school_df, geometry="geometry", crs="EPSG:26986")

# Convert to lat/lon (WGS84)
school_gdf = school_gdf.to_crs(epsg=4326)

# Extract latitude & longitude
school_gdf["longitude"] = school_gdf.geometry.x
school_gdf["latitude"] = school_gdf.geometry.y

school_gdf[["latitude", "longitude"]].head()
```

```python
school_gdf.columns
```

```python
school_gdf[['OBJECTID', 'SCHID', 'NAME', 'ADDRESS', 'TOWN_MAIL', 'TOWN', 'ZIPCODE',
       'PHONE', 'GRADES', 'TYPE', 'TYPE_DESC', 'DIST_NAME', 'longitude', 'latitude']].to_csv('../../streamlit-app/school_geocoded.csv', index = None)
```
